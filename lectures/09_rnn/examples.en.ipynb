# Lecture 09 -- experiment: RNN, LSTM, GRU

Companion material to [`lecture.en.md`](lecture.en.md). Four experiments:
(A) unrolling over time -- a manual cell vs. `nn.RNN`; (B) the vanishing
gradient -- measured, not assumed; (C) an order-dependent token task;
(D) LSTM/GRU gates, shapes, parameters, time.

```python
import time

import matplotlib.pyplot as plt
import torch
import torch.nn as nn

torch.manual_seed(0)
torch.set_num_threads(4)  # BLAS thread stability under jupyter/ipykernel (lesson from module 07)
device = torch.device("cpu")
print(f"Using device: {device}")
```

## A. Unrolling an RNN over time: a manual cell vs. `nn.RNN`

The same numerical example as in `lecture.en.md`: $d=2$, $d_h=2$, $T=3$.

```python
d_in, d_h, T = 2, 2, 3

Wx = torch.tensor([[0.5, -0.3], [0.1, 0.8]])
Wh = torch.tensor([[0.2, 0.0], [-0.1, 0.4]])
b = torch.tensor([0.1, -0.1])

X = torch.tensor(
    [
        [1.0, 0.0],
        [0.0, 1.0],
        [1.0, 1.0],
    ]
)
h0 = torch.zeros(d_h)


def manual_rnn_forward(X, Wx, Wh, b, h0):
    """By hand, with basic operations -- the SAME function f(x_t, h_{t-1}) called T times."""
    h_prev = h0
    states = []
    for t in range(X.shape[0]):
        h_t = torch.tanh(Wx @ X[t] + Wh @ h_prev + b)
        states.append(h_t)
        h_prev = h_t
    return torch.stack(states)


manual_states = manual_rnn_forward(X, Wx, Wh, b, h0)
print("manual hidden states h_1, h_2, h_3:")
print(manual_states)
```

```python
rnn = nn.RNN(input_size=d_in, hidden_size=d_h, num_layers=1, nonlinearity="tanh", batch_first=True)
with torch.no_grad():
    # PyTorch uses TWO separate biases (b_ih, b_hh), summed -- see
    # lecture.en.md. We set b_ih=b, b_hh=0, so their sum equals our single b.
    rnn.weight_ih_l0.copy_(Wx)
    rnn.weight_hh_l0.copy_(Wh)
    rnn.bias_ih_l0.copy_(b)
    rnn.bias_hh_l0.zero_()

out, h_n = rnn(X.unsqueeze(0), h0.view(1, 1, d_h))
print("nn.RNN hidden states:")
print(out.squeeze(0))

match = torch.allclose(manual_states, out.squeeze(0), atol=1e-6)
print(f"\nmanual == nn.RNN (tolerance atol=1e-6): {match}")
assert match
```

### Effect of changing an earlier sequence element on later states

We change ONLY $x_1$ and check which states change.

```python
X_modified = X.clone()
X_modified[0] = torch.tensor([-1.0, -1.0])  # change ONLY the first input

states_original = manual_rnn_forward(X, Wx, Wh, b, h0)
states_modified = manual_rnn_forward(X_modified, Wx, Wh, b, h0)

for t in range(T):
    changed = not torch.allclose(states_original[t], states_modified[t], atol=1e-6)
    print(f"h_{t + 1}: changed after modifying x_1 = {changed}")

print(
    "\nAll three states changed -- changing an EARLIER input affects EVERY "
    "later state (because every h_t depends on h_{t-1}, which depends on "
    "h_{t-2}, ... all the way back to x_1). This is a direct consequence "
    "of the recurrence, not a coincidence."
)
```

## B. The vanishing gradient -- measured, not assumed

The norm of $\partial L/\partial x_1$ (loss = sum of squares of the last
hidden state) as a function of sequence length $T$, for RNN, LSTM, GRU --
**the same** $d_h$, **the same** random initialization in every
comparison (different seeds checked separately).

```python
def grad_wrt_first_input(model_type, T, d_in=4, d_h=8, seed=0):
    torch.manual_seed(seed)
    if model_type == "rnn":
        model = nn.RNN(d_in, d_h, batch_first=True)
    elif model_type == "lstm":
        model = nn.LSTM(d_in, d_h, batch_first=True)
    elif model_type == "gru":
        model = nn.GRU(d_in, d_h, batch_first=True)

    X_seq = torch.randn(1, T, d_in, requires_grad=True)
    out = model(X_seq)
    final_h = out[0][:, -1, :]
    loss = final_h.pow(2).sum()
    loss.backward()

    grad_x1_norm = X_seq.grad[0, 0].norm().item()
    grad_Whh_norm = model.weight_hh_l0.grad.norm().item()
    return grad_x1_norm, grad_Whh_norm


seq_lengths = [2, 5, 10, 20, 40, 80]
results = {"rnn": [], "lstm": [], "gru": []}
for T_val in seq_lengths:
    for name in results:
        g_x1, _ = grad_wrt_first_input(name, T_val)
        results[name].append(g_x1)

print(f"{'T':>4} {'RNN |grad x1|':>15} {'LSTM |grad x1|':>16} {'GRU |grad x1|':>15}")
for i, T_val in enumerate(seq_lengths):
    print(
        f"{T_val:>4} {results['rnn'][i]:>15.3e} {results['lstm'][i]:>16.3e} "
        f"{results['gru'][i]:>15.3e}"
    )
```

```python
fig, ax = plt.subplots(figsize=(6.5, 4.5))
for name, label in [("rnn", "RNN"), ("lstm", "LSTM"), ("gru", "GRU")]:
    ax.plot(seq_lengths, results[name], marker="o", label=label)
ax.set_yscale("log")
ax.set_xlabel("sequence length T")
ax.set_ylabel("norm of $\\partial L/\\partial x_1$ (log scale)")
ax.set_title("Vanishing gradient with respect to the first input")
ax.legend()
plt.tight_layout()
plt.show()

print(
    "All three architectures vanish to practically zero at a sufficiently "
    "long sequence -- LSTM/GRU do NOT eliminate the problem, they slow it down."
)
```

**Do LSTM/GRU always have a larger gradient than RNN?** We check across
several seeds at $T=40$ -- we do not assume the answer in advance.

```python
print("seed   RNN          LSTM         GRU")
for seed in [0, 1, 2, 3, 4]:
    g_rnn, _ = grad_wrt_first_input("rnn", 40, seed=seed)
    g_lstm, _ = grad_wrt_first_input("lstm", 40, seed=seed)
    g_gru, _ = grad_wrt_first_input("gru", 40, seed=seed)
    print(f"{seed:>4}   {g_rnn:.3e}   {g_lstm:.3e}   {g_gru:.3e}")
```

### A different quantity: the gradient with respect to the shared matrix $W_h$

This is **not** the same as the gradient with respect to $x_1$ -- it sums
contributions from ALL time steps, so even when individual, distant
contributions vanish, nearby steps still give a healthy signal.

```python
print(f"{'T':>4} {'RNN |grad W_hh|':>17} {'LSTM |grad W_hh|':>18} {'GRU |grad W_hh|':>17}")
for T_val in seq_lengths:
    _, w_rnn = grad_wrt_first_input("rnn", T_val)
    _, w_lstm = grad_wrt_first_input("lstm", T_val)
    _, w_gru = grad_wrt_first_input("gru", T_val)
    print(f"{T_val:>4} {w_rnn:>17.3e} {w_lstm:>18.3e} {w_gru:>17.3e}")

print(
    "\nUnlike the gradient with respect to x_1 (which vanishes by orders of "
    "magnitude), the gradient with respect to W_hh does NOT decline "
    "systematically with T -- this is a different quantity, with different "
    "behavior. They are not interchangeable."
)
```

## C. A task depending on token order

Sequences = permutations of THE SAME set of 6 tokens (an identical
bag-of-tokens in every sequence) -- the label depends SOLELY on whether
token 0 appears before token 1. A baseline based on averaging embeddings
cannot have access to this information (the mean is invariant to
permutations of the input).

```python
VOCAB_SIZE, SEQ_LEN = 6, 6


def make_order_dataset(n, seed):
    g = torch.Generator().manual_seed(seed)
    seqs = torch.stack([torch.randperm(VOCAB_SIZE, generator=g) for _ in range(n)])
    pos0 = (seqs == 0).float().argmax(dim=1)
    pos1 = (seqs == 1).float().argmax(dim=1)
    labels = (pos0 < pos1).long()
    return seqs, labels


X_train, y_train = make_order_dataset(800, seed=0)
X_val, y_val = make_order_dataset(200, seed=1)
X_test, y_test = make_order_dataset(200, seed=2)

for seq in X_train[:20]:
    assert sorted(seq.tolist()) == list(range(VOCAB_SIZE)), "bag-of-tokens is not identical!"
print(f"train label balance: {y_train.float().mean():.3f} (should be about 0.5)")
print("OK: every sequence is a permutation of the same set of tokens.")
```

```python
d_embed, d_hidden = 8, 16


class MeanPoolBaseline(nn.Module):
    def __init__(self):
        super().__init__()
        self.embed = nn.Embedding(VOCAB_SIZE, d_embed)
        self.fc = nn.Linear(d_embed, 2)

    def forward(self, x):
        e = self.embed(x)
        pooled = e.mean(dim=1)  # invariant to permutations of x
        return self.fc(pooled)


class SeqClassifier(nn.Module):
    def __init__(self, cell):
        super().__init__()
        self.embed = nn.Embedding(VOCAB_SIZE, d_embed)
        self.rnn = cell(d_embed, d_hidden, batch_first=True)
        self.fc = nn.Linear(d_hidden, 2)

    def forward(self, x):
        e = self.embed(x)
        out, _ = self.rnn(e)
        return self.fc(out[:, -1, :])


def count_params(model):
    return sum(p.numel() for p in model.parameters())


def train_eval(model, epochs=30, lr=0.01, batch_size=32, seed=0):
    torch.manual_seed(seed)
    opt = torch.optim.Adam(model.parameters(), lr=lr)
    loss_fn = nn.CrossEntropyLoss()
    n = X_train.shape[0]
    start = time.time()
    for _ in range(epochs):
        perm = torch.randperm(n)
        for i in range(0, n, batch_size):
            idx = perm[i : i + batch_size]
            opt.zero_grad()
            loss = loss_fn(model(X_train[idx]), y_train[idx])
            loss.backward()
            opt.step()
    train_time = time.time() - start
    model.eval()
    with torch.no_grad():
        val_acc = (model(X_val).argmax(1) == y_val).float().mean().item()
        test_acc = (model(X_test).argmax(1) == y_test).float().mean().item()
    return val_acc, test_acc, train_time


print(f"{'model':30s} {'params':>8s} {'val_acc':>8s} {'test_acc':>9s} {'time':>7s}")
for name, build in [
    ("mean-pool (no order)", lambda: MeanPoolBaseline()),
    ("vanilla RNN", lambda: SeqClassifier(nn.RNN)),
    ("GRU", lambda: SeqClassifier(nn.GRU)),
]:
    m = build()
    n_params = count_params(m)
    val_acc, test_acc, t = train_eval(m)
    print(f"{name:30s} {n_params:>8d} {val_acc:>8.3f} {test_acc:>9.3f} {t:>6.2f}s")
```

**Interpretation.** The order-blind baseline reaches close-to-chance
accuracy -- **not** because of lower capacity (it actually has fewer
parameters, which is not itself the cause), but because averaging
embeddings is **mathematically invariant** to the order of the input
tokens -- the information that decides the task is **unavailable to it in
principle**, regardless of the training budget. RNN and GRU, having
access to order, solve the task almost perfectly.

## D. LSTM and GRU: gates, shapes, parameters, time

```python
torch.manual_seed(0)
d_in_d, d_h_d, T_d = 3, 2, 4
X_d = torch.randn(1, T_d, d_in_d)

lstm = nn.LSTM(d_in_d, d_h_d, batch_first=True)
out_lstm, (hn_lstm, cn_lstm) = lstm(X_d)
print(f"LSTM: input {tuple(X_d.shape)} -> output (all h_t) {tuple(out_lstm.shape)}")
print(f"  h_n (last hidden state): {tuple(hn_lstm.shape)}")
print(f"  c_n (last cell state): {tuple(cn_lstm.shape)}")

gru = nn.GRU(d_in_d, d_h_d, batch_first=True)
out_gru, hn_gru = gru(X_d)
print(f"\nGRU: input {tuple(X_d.shape)} -> output {tuple(out_gru.shape)}")
print(f"  h_n: {tuple(hn_gru.shape)} (NO separate cell state -- GRU does not have one)")
```

```python
# manual verification of the LSTM gate equations against nn.LSTM
W_ih, W_hh = lstm.weight_ih_l0, lstm.weight_hh_l0
b_ih, b_hh = lstm.bias_ih_l0, lstm.bias_hh_l0
W_ii, W_if, W_ig, W_io = W_ih.chunk(4, dim=0)
W_hi, W_hf, W_hg, W_ho = W_hh.chunk(4, dim=0)
b_ii, b_if, b_ig, b_io = b_ih.chunk(4, dim=0)
b_hi, b_hf, b_hg, b_ho = b_hh.chunk(4, dim=0)

h_prev, c_prev = torch.zeros(d_h_d), torch.zeros(d_h_d)
manual_h, manual_c = [], []
for t in range(T_d):
    x_t = X_d[0, t]
    i_t = torch.sigmoid(W_ii @ x_t + b_ii + W_hi @ h_prev + b_hi)
    f_t = torch.sigmoid(W_if @ x_t + b_if + W_hf @ h_prev + b_hf)
    g_t = torch.tanh(W_ig @ x_t + b_ig + W_hg @ h_prev + b_hg)
    o_t = torch.sigmoid(W_io @ x_t + b_io + W_ho @ h_prev + b_ho)
    c_t = f_t * c_prev + i_t * g_t
    h_t = o_t * torch.tanh(c_t)
    manual_h.append(h_t)
    manual_c.append(c_t)
    h_prev, c_prev = h_t, c_t

match_h = torch.allclose(torch.stack(manual_h), out_lstm.squeeze(0), atol=1e-6)
match_c = torch.allclose(c_prev, cn_lstm.squeeze(), atol=1e-6)
print(f"manual LSTM gate equations == nn.LSTM: h: {match_h}, c: {match_c}")
assert match_h and match_c
```

```python
# parameter count and time: RNN vs LSTM vs GRU, the same d_in/d_h
d_in_cmp, d_h_cmp, T_cmp, batch_cmp = 32, 64, 50, 64
X_cmp = torch.randn(batch_cmp, T_cmp, d_in_cmp)

print(f"{'model':8s} {'params':>10s} {'ratio to RNN':>14s} {'forward time (ms)':>19s}")
rnn_params = None
for name, cls in [("RNN", nn.RNN), ("GRU", nn.GRU), ("LSTM", nn.LSTM)]:
    torch.manual_seed(0)
    m = cls(d_in_cmp, d_h_cmp, batch_first=True)
    n_params = sum(p.numel() for p in m.parameters())
    if name == "RNN":
        rnn_params = n_params
    start = time.time()
    with torch.no_grad():
        for _ in range(20):
            m(X_cmp)
    elapsed_ms = (time.time() - start) / 20 * 1000
    print(f"{name:8s} {n_params:>10d} {n_params / rnn_params:>13.1f}x {elapsed_ms:>18.2f}")
```

## Summary

- The manual RNN cell agrees with `nn.RNN` within numerical tolerance
  (`atol=1e-6`) -- an RNN is the repeated application of the same
  function, verified, not just declared.
- The gradient with respect to the first input vanishes exponentially
  with sequence length for all three architectures -- LSTM/GRU slow this
  phenomenon down, they do not eliminate it. The gradient with respect to
  the shared weights $W_h$ behaves completely differently.
- A baseline with no access to order is mathematically limited to chance
  level on a task constructed so that order is the only decisive
  information -- RNN/GRU solve it almost perfectly.
- LSTM has exactly $4\times$ the parameters of an RNN, GRU exactly
  $3\times$ -- verified, not approximated.
